```python
from sklearn.preprocessing import StandardScaler


# Select numeric features
X = df[['Age', 'Annual_Income', 'Spending_score']]

# Apply standardization
scaler = StandardScaler()
X_standardized = scaler.fit_transform(X)

# Convert back to DataFrame (for readability & plotting)
X_standardized_df = pd.DataFrame(
    X_standardized,
    columns=['Age_std', 'Annual_Income_std', 'Spending_score_std']
)

X_standardized_df.head()
```

```python
X.mean()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
sns.scatterplot(
    x='Annual_Income_std',
    y='Spending_score_std',
    data=X_standardized_df,
    alpha=0.6
)
plt.title('After Standardization (StandardScaler)')
plt.show()
```

```python
X_standardized_df[['Annual_Income_std', 'Spending_score_std']].std()
```

```python
X_standardized_df[['Annual_Income_std', 'Spending_score_std']].mean()
```

```python
from sklearn.preprocessing import MinMaxScaler
import pandas as pd

X = df[['Annual_Income', 'Spending_score']]

scaler = MinMaxScaler()
X_minmax = scaler.fit_transform(X)

X_minmax_df = pd.DataFrame(
    X_minmax,
    columns=['Annual_Income_scaled', 'Spending_score_scaled']
)

X_minmax_df.head()
```

```python
X.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))
sns.scatterplot(
    x='Annual_Income_scaled',
    y='Spending_score_scaled',
    data=X_minmax_df,
    alpha=0.6
)
plt.title('After Min-Max Scaling')
plt.xlabel('Annual Income (0–1)')
plt.ylabel('Spending Score (0–1)')
plt.show()
```
