## 📈 Simple Linear Regression Analysis


# 🎯 Objective

The objective of this notebook is to understand and implement 
Simple Linear Regression to model the relationship between 
an independent variable (feature) and a dependent variable (target).

We aim to:
- Visualize the data
- Train a regression model
- Evaluate model performance
- Interpret results


```python
from sklearn.datasets import fetch_california_housing
import pandas as pd

housing = fetch_california_housing()

df = pd.DataFrame(housing.data, columns=housing.feature_names)
df["Price"] = housing.target

print(df.head())
print(df.info())

# Define features and target
X = df.drop("Price", axis=1)
y = df["Price"]

print("X shape:", X.shape)
print("y shape:", y.shape)


```

```python
import pandas as pd
import numpy as np

# Combine features + target
df_full = pd.concat([X, y], axis=1)

# Compute IQR
Q1 = df_full.quantile(0.25)
Q3 = df_full.quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Filter outliers
df_clean = df_full[
    ~((df_full < lower_bound) | (df_full > upper_bound)).any(axis=1)
]

print("Original shape :", df_full.shape)
print("After cleaning :", df_clean.shape)
```

```python
X_clean = df_clean.drop("Price", axis=1)
y_clean = df_clean["Price"]
```

# ✂ Train–Test Split

The dataset is divided into:

- Training Set → Used to train the model
- Testing Set → Used to evaluate performance

This helps assess model generalization.


```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_clean, y_clean,
    test_size=0.2,
    random_state=42
)

print("Training set:", X_train.shape)
print("Testing set :", X_test.shape)
```

# 🤖 Model Training – Linear Regression

We train a Simple Linear Regression model 
to learn the relationship between X and y.


```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

# 📊 Model Evaluation Metrics

We evaluate the regression model using:

- **MAE (Mean Absolute Error)** → Average absolute prediction error
- **RMSE (Root Mean Squared Error)** → Penalizes large errors
- **R² (R-squared)** → Proportion of variance explained


```python
from sklearn.metrics import mean_absolute_error, root_mean_squared_error, r2_score

print("📊 Metrics After Proper Cleaning")
print("MAE :", mean_absolute_error(y_test, y_pred))
print("RMSE:", root_mean_squared_error(y_test, y_pred))
print("R²  :", r2_score(y_test, y_pred))
```

# 📈 Regression Line

The regression line represents the best-fit linear relationship 
between the feature and the target variable.


```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.regplot(
    x=y_test,
    y=y_pred,
    scatter_kws={"alpha": 0.5}  # makes points slightly transparent
)

plt.xlabel("Actual Prices")
plt.ylabel("Predicted Prices")
plt.title("Actual vs Predicted with Regression Line")
plt.show()
```

```python
residuals = y_test - y_pred

sns.scatterplot(x=y_pred, y=residuals)
plt.axhline(0)
plt.title("Residual Plot")
plt.show()
```
